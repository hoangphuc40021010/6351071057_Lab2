# Câu 4

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
from sklearn.metrics import precision_score, recall_score, f1_score
import warnings
warnings.filterwarnings('ignore')

# In thông tin sinh viên
print("="*60)
print("6351071057 - Lê Hoàng Phúc")
print("PHÂN TÍCH DỮ LIỆU MUSHROOM")
print("="*60)

# ============================================================================
# NHẬP VÀ KHẢO SÁT DỮ LIỆU
# ============================================================================
print("\n" + "="*60)
print("NHẬP VÀ KHẢO SÁT DỮ LIỆU")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# Đọc dữ liệu - thử nhiều cách khác nhau
try:
    # Thử đọc với encoding mặc định
    df = pd.read_csv('mushrooms.csv')
    print("\n✓ Đã đọc dữ liệu từ file mushrooms.csv")
except:
    try:
        # Thử với encoding khác
        df = pd.read_csv('mushrooms.csv', encoding='latin-1')
        print("\n✓ Đã đọc dữ liệu từ file mushrooms.csv (encoding: latin-1)")
    except:
        try:
            # Thử không có header
            df = pd.read_csv('mushrooms.csv', header=None)
            # Đặt tên cột thủ công
            df.columns = ['class', 'cap-shape', 'cap-surface', 'cap-color', 'bruises',
                         'odor', 'gill-attachment', 'gill-spacing', 'gill-size', 'gill-color',
                         'stalk-shape', 'stalk-root', 'stalk-surface-above-ring',
                         'stalk-surface-below-ring', 'stalk-color-above-ring',
                         'stalk-color-below-ring', 'veil-type', 'veil-color',
                         'ring-number', 'ring-type', 'spore-print-color', 'population', 'habitat']
            print("\n✓ Đã đọc dữ liệu từ file mushroom4.csv (không có header)")
        except Exception as e:
            print(f"\n✗ Lỗi đọc file: {e}")
            print("→ Vui lòng kiểm tra lại file mushroom4.csv")
            raise

print("\n1. Dữ liệu ban đầu:")
print("6351071057 - Lê Hoàng Phúc")
print(f"Kích thước: {df.shape}")
print(f"\nTên các cột:\n{df.columns.tolist()}")
print(f"\n10 dòng đầu tiên:\n{df.head(10)}")

# Kiểm tra kiểu dữ liệu
print("\n2. Kiểu dữ liệu:")
print("6351071057 - Lê Hoàng Phúc")
print(df.dtypes)

# Kiểm tra dữ liệu trống
print("\n3. Kiểm tra dữ liệu trống:")
print("6351071057 - Lê Hoàng Phúc")
null_counts = df.isnull().sum()
print(f"Tổng số giá trị null trong mỗi cột:")
print(null_counts)
print(f"\nTổng số dòng có giá trị null: {df.isnull().any(axis=1).sum()}")

# Kiểm tra các giá trị ? (missing values trong dataset mushroom)
print("\n4. Kiểm tra giá trị '?' (missing values):")
print("6351071057 - Lê Hoàng Phúc")
for col in df.columns:
    question_count = (df[col] == '?').sum()
    if question_count > 0:
        print(f"{col}: {question_count} giá trị '?'")

# Thay thế '?' bằng mode hoặc xóa dòng
if (df == '?').any().any():
    print("\n→ Tìm thấy giá trị '?', đang xử lý...")
    # Xóa các dòng có '?'
    df_cleaned = df[~(df == '?').any(axis=1)]
    print(f"→ Đã xóa {df.shape[0] - df_cleaned.shape[0]} dòng chứa '?'")
else:
    df_cleaned = df.copy()
    print("\n→ Không có giá trị '?' trong dữ liệu")

# Xóa các dòng có dữ liệu trống (null)
df_cleaned = df_cleaned.dropna()
print(f"\nKích thước sau khi xóa dòng trống: {df_cleaned.shape}")
if df.shape[0] != df_cleaned.shape[0]:
    print(f"Đã xóa tổng cộng {df.shape[0] - df_cleaned.shape[0]} dòng")
else:
    print("Không có dòng nào bị xóa thêm")

df = df_cleaned.copy()

# Phân bố nhãn class
print("\n5. Phân bố nhãn 'class':")
print("6351071057 - Lê Hoàng Phúc")
print(df['class'].value_counts())
print(f"\nTỷ lệ:")
print(df['class'].value_counts(normalize=True))

# Visualize target distribution
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
df['class'].value_counts().plot(kind='bar', color=['lightgreen', 'lightcoral'])
plt.title('Phân bố nhãn Class\n6351071057 - Lê Hoàng Phúc')
plt.xlabel('Class (e: Edible, p: Poisonous)')
plt.ylabel('Số lượng')
plt.xticks(rotation=0)

plt.subplot(1, 2, 2)
class_counts = df['class'].value_counts()
labels = ['Edible (e)' if idx == 'e' else 'Poisonous (p)' for idx in class_counts.index]
colors = ['lightgreen' if idx == 'e' else 'lightcoral' for idx in class_counts.index]
plt.pie(class_counts, autopct='%1.1f%%',
        colors=colors,
        labels=labels)
plt.title('Tỷ lệ phân bố Class\n6351071057 - Lê Hoàng Phúc')

plt.tight_layout()
plt.show()

# Kiểm tra giá trị unique của mỗi cột
print("\n6. Số giá trị unique của mỗi thuộc tính:")
print("6351071057 - Lê Hoàng Phúc")
for col in df.columns:
    unique_vals = sorted(df[col].unique())
    print(f"{col}: {df[col].nunique()} giá trị - {unique_vals[:10]}{'...' if len(unique_vals) > 10 else ''}")
```

```python
# ============================================================================
# YÊU CẦU 4: TÁCH DỮ LIỆU THÀNH THUỘC TÍNH VÀ NHÃN
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 4: TÁCH DỮ LIỆU THÀNH THUỘC TÍNH VÀ NHÃN")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

X = df.drop('class', axis=1)  # Thuộc tính
y = df['class']  # Nhãn quyết định

print(f"\nKích thước thuộc tính X: {X.shape}")
print(f"Kích thước nhãn y: {y.shape}")
print(f"\nCác thuộc tính ({len(X.columns)} cột):\n{list(X.columns)}")
print(f"\nPhân bố nhãn class:")
print("6351071057 - Lê Hoàng Phúc")
print(y.value_counts().sort_index())
```

# Câu 5

```python
# ============================================================================
# YÊU CẦU 5: CHUYỂN ĐỔI ONE-HOT ENCODING
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 5: CHUYỂN ĐỔI ONE-HOT ENCODING")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# Kiểm tra các cột không phải số
non_numeric_cols = X.select_dtypes(include=['object']).columns.tolist()
print(f"\nCác cột không phải số: {len(non_numeric_cols)} cột")
print(f"Tổng số cột cần chuyển đổi: {len(non_numeric_cols)}")

# Áp dụng one-hot encoding
print("\n→ Đang thực hiện one-hot encoding...")
X_encoded = pd.get_dummies(X, columns=non_numeric_cols, drop_first=True)

print(f"\nKích thước trước encoding: {X.shape}")
print(f"Kích thước sau one-hot encoding: {X_encoded.shape}")
print(f"\nSố cột đã tăng từ {X.shape[1]} lên {X_encoded.shape[1]}")
print(f"\nMột số cột sau encoding (20 cột đầu):")
for i, col in enumerate(X_encoded.columns[:20]):
    print(f"  {i+1}. {col}")
if X_encoded.shape[1] > 20:
    print(f"  ... và {X_encoded.shape[1] - 20} cột khác")
```

# câu 6

```python
# ============================================================================
# YÊU CẦU 6: CHIA DỮ LIỆU TRAIN/TEST (70/30)
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 6: CHIA DỮ LIỆU TRAIN/TEST (70/30)")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.3, random_state=42, stratify=y
)

print(f"\nKích thước tập huấn luyện: {X_train.shape}")
print(f"Kích thước tập kiểm thử: {X_test.shape}")
print(f"\nPhân bố nhãn trong tập huấn luyện:")
print("6351071057 - Lê Hoàng Phúc")
print(y_train.value_counts().sort_index())
print(f"\nTỷ lệ: \n{y_train.value_counts(normalize=True).sort_index()}")
print(f"\nPhân bố nhãn trong tập kiểm thử:")
print(y_test.value_counts().sort_index())
print(f"\nTỷ lệ: \n{y_test.value_counts(normalize=True).sort_index()}")
```

# câu 7

```python
# ============================================================================
# YÊU CẦU 7: CÂY QUYẾT ĐỊNH ID3 (ENTROPY)
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 7: CÂY QUYẾT ĐỊNH ID3 (ENTROPY)")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# ID3 sử dụng entropy làm tiêu chí
print("\n→ Đang huấn luyện mô hình ID3...")
id3_model = DecisionTreeClassifier(criterion='entropy', random_state=42, max_depth=5)
id3_model.fit(X_train, y_train)

# Dự đoán
y_pred_id3 = id3_model.predict(X_test)

# Ma trận nhầm lẫn
cm_id3 = confusion_matrix(y_test, y_pred_id3)
print("\nMa trận nhầm lẫn ID3:")
print("6351071057 - Lê Hoàng Phúc")
print(cm_id3)

# Độ chính xác
accuracy_id3 = accuracy_score(y_test, y_pred_id3)
print(f"\nĐộ chính xác ID3: {accuracy_id3:.4f} ({accuracy_id3*100:.2f}%)")

# Classification report
print("\nBáo cáo phân loại ID3:")
print("6351071057 - Lê Hoàng Phúc")
print(classification_report(y_test, y_pred_id3, target_names=['Edible (e)', 'Poisonous (p)']))

# Visualize confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm_id3, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Edible (e)', 'Poisonous (p)'],
            yticklabels=['Edible (e)', 'Poisonous (p)'])
plt.title('Ma trận nhầm lẫn - Cây ID3\n6351071057 - Lê Hoàng Phúc')
plt.ylabel('Nhãn thực tế')
plt.xlabel('Nhãn dự đoán')
plt.tight_layout()
plt.show()

# Visualize tree
plt.figure(figsize=(20, 12))
plot_tree(id3_model, feature_names=X_encoded.columns,
          class_names=['Edible', 'Poisonous'],
          filled=True, rounded=True, fontsize=7)
plt.title('Cây quyết định ID3 (max_depth=5)\n6351071057 - Lê Hoàng Phúc', fontsize=14)
plt.tight_layout()
plt.show()

# Feature importance
feature_importance_id3 = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Importance': id3_model.feature_importances_
}).sort_values('Importance', ascending=False)

print("\nĐộ quan trọng của các thuộc tính (ID3) - Top 10:")
print("6351071057 - Lê Hoàng Phúc")
print(feature_importance_id3.head(10))

# Visualize feature importance
plt.figure(figsize=(10, 6))
top_features = feature_importance_id3.head(10)
plt.barh(top_features['Feature'], top_features['Importance'], color='skyblue')
plt.xlabel('Độ quan trọng')
plt.title('Top 10 thuộc tính quan trọng nhất (ID3)\n6351071057 - Lê Hoàng Phúc')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

# câu 8

```python
# ============================================================================
# YÊU CẦU 8: CÂY QUYẾT ĐỊNH CART (GINI)
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 8: CÂY QUYẾT ĐỊNH CART (GINI)")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# CART sử dụng gini làm tiêu chí
print("\n→ Đang huấn luyện mô hình CART...")
cart_model = DecisionTreeClassifier(criterion='gini', random_state=42, max_depth=5)
cart_model.fit(X_train, y_train)

# Dự đoán
y_pred_cart = cart_model.predict(X_test)

# Ma trận nhầm lẫn
cm_cart = confusion_matrix(y_test, y_pred_cart)
print("\nMa trận nhầm lẫn CART:")
print("6351071057 - Lê Hoàng Phúc")
print(cm_cart)

# Độ chính xác
accuracy_cart = accuracy_score(y_test, y_pred_cart)
print(f"\nĐộ chính xác CART: {accuracy_cart:.4f} ({accuracy_cart*100:.2f}%)")

# Classification report
print("\nBáo cáo phân loại CART:")
print("6351071057 - Lê Hoàng Phúc")
print(classification_report(y_test, y_pred_cart, target_names=['Edible (e)', 'Poisonous (p)']))

# Visualize confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm_cart, annot=True, fmt='d', cmap='Greens',
            xticklabels=['Edible (e)', 'Poisonous (p)'],
            yticklabels=['Edible (e)', 'Poisonous (p)'])
plt.title('Ma trận nhầm lẫn - Cây CART\n6351071057 - Lê Hoàng Phúc')
plt.ylabel('Nhãn thực tế')
plt.xlabel('Nhãn dự đoán')
plt.tight_layout()
plt.show()

# Visualize tree
plt.figure(figsize=(20, 12))
plot_tree(cart_model, feature_names=X_encoded.columns,
          class_names=['Edible', 'Poisonous'],
          filled=True, rounded=True, fontsize=7)
plt.title('Cây quyết định CART (max_depth=5)\n6351071057 - Lê Hoàng Phúc', fontsize=14)
plt.tight_layout()
plt.show()

# Feature importance
feature_importance_cart = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Importance': cart_model.feature_importances_
}).sort_values('Importance', ascending=False)

print("\nĐộ quan trọng của các thuộc tính (CART) - Top 10:")
print("6351071057 - Lê Hoàng Phúc")
print(feature_importance_cart.head(10))

# Visualize feature importance
plt.figure(figsize=(10, 6))
top_features = feature_importance_cart.head(10)
plt.barh(top_features['Feature'], top_features['Importance'], color='lightgreen')
plt.xlabel('Độ quan trọng')
plt.title('Top 10 thuộc tính quan trọng nhất (CART)\n6351071057 - Lê Hoàng Phúc')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

# câu 9

```python
# ============================================================================
# YÊU CẦU 9: NAÏVE BAYES
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 9: NAÏVE BAYES")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

print("\n→ Đang huấn luyện mô hình Naïve Bayes...")
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)

# Dự đoán
y_pred_nb = nb_model.predict(X_test)

# Ma trận nhầm lẫn
cm_nb = confusion_matrix(y_test, y_pred_nb)
print("\nMa trận nhầm lẫn Naïve Bayes:")
print("6351071057 - Lê Hoàng Phúc")
print(cm_nb)

# Độ chính xác
accuracy_nb = accuracy_score(y_test, y_pred_nb)
print(f"\nĐộ chính xác Naïve Bayes: {accuracy_nb:.4f} ({accuracy_nb*100:.2f}%)")

# Classification report
print("\nBáo cáo phân loại Naïve Bayes:")
print("6351071057 - Lê Hoàng Phúc")
print(classification_report(y_test, y_pred_nb, target_names=['Edible (e)', 'Poisonous (p)']))

# Visualize confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm_nb, annot=True, fmt='d', cmap='Purples',
            xticklabels=['Edible (e)', 'Poisonous (p)'],
            yticklabels=['Edible (e)', 'Poisonous (p)'])
plt.title('Ma trận nhầm lẫn - Naïve Bayes\n6351071057 - Lê Hoàng Phúc')
plt.ylabel('Nhãn thực tế')
plt.xlabel('Nhãn dự đoán')
plt.tight_layout()
plt.show()
```

```python
# ============================================================================
# YÊU CẦU 10: SO SÁNH CÁC MÔ HÌNH
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 10: SO SÁNH CÁC MÔ HÌNH")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# Bảng so sánh
comparison_df = pd.DataFrame({
    'Mô hình': ['ID3 (Entropy)', 'CART (Gini)', 'Naïve Bayes'],
    'Độ chính xác': [accuracy_id3, accuracy_cart, accuracy_nb],
    'Độ chính xác (%)': [accuracy_id3*100, accuracy_cart*100, accuracy_nb*100]
})

print("\nBảng so sánh độ chính xác:")
print("6351071057 - Lê Hoàng Phúc")
print(comparison_df)

# Visualize comparison
plt.figure(figsize=(12, 6))

# Bar chart
plt.subplot(1, 2, 1)
bars = plt.bar(comparison_df['Mô hình'], comparison_df['Độ chính xác'],
               color=['skyblue', 'lightgreen', 'plum'])
plt.ylabel('Độ chính xác')
plt.title('So sánh độ chính xác các mô hình\n6351071057 - Lê Hoàng Phúc')
plt.ylim(0, 1.1)
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.4f}\n({height*100:.2f}%)',
             ha='center', va='bottom')

# Line chart
plt.subplot(1, 2, 2)
plt.plot(comparison_df['Mô hình'], comparison_df['Độ chính xác'],
         marker='o', linewidth=2, markersize=10, color='coral')
plt.ylabel('Độ chính xác')
plt.title('Xu hướng độ chính xác\n6351071057 - Lê Hoàng Phúc')
plt.ylim(0, 1.1)
plt.grid(True, alpha=0.3)
for i, (model, acc) in enumerate(zip(comparison_df['Mô hình'], comparison_df['Độ chính xác'])):
    plt.text(i, acc + 0.02, f'{acc:.4f}', ha='center')

plt.tight_layout()
plt.show()

# So sánh chi tiết với các metrics khác
detailed_comparison = pd.DataFrame({
    'Mô hình': ['ID3 (Entropy)', 'CART (Gini)', 'Naïve Bayes'],
    'Accuracy': [
        accuracy_score(y_test, y_pred_id3),
        accuracy_score(y_test, y_pred_cart),
        accuracy_score(y_test, y_pred_nb)
    ],
    'Precision': [
        precision_score(y_test, y_pred_id3, average='weighted'),
        precision_score(y_test, y_pred_cart, average='weighted'),
        precision_score(y_test, y_pred_nb, average='weighted')
    ],
    'Recall': [
        recall_score(y_test, y_pred_id3, average='weighted'),
        recall_score(y_test, y_pred_cart, average='weighted'),
        recall_score(y_test, y_pred_nb, average='weighted')
    ],
    'F1-Score': [
        f1_score(y_test, y_pred_id3, average='weighted'),
        f1_score(y_test, y_pred_cart, average='weighted'),
        f1_score(y_test, y_pred_nb, average='weighted')
    ]
})

print("\nSo sánh chi tiết các metrics:")
print("6351071057 - Lê Hoàng Phúc")
print(detailed_comparison)

# Visualize detailed comparison
plt.figure(figsize=(12, 6))
x = np.arange(len(detailed_comparison['Mô hình']))
width = 0.2

metrics = ['Accuracy', 'Precision', 'Recall', 'F1-Score']
colors = ['skyblue', 'lightgreen', 'lightcoral', 'plum']

for i, metric in enumerate(metrics):
    plt.bar(x + i*width, detailed_comparison[metric], width,
            label=metric, color=colors[i])

plt.xlabel('Mô hình')
plt.ylabel('Giá trị')
plt.title('So sánh chi tiết các metrics\n6351071057 - Lê Hoàng Phúc')
plt.xticks(x + width*1.5, detailed_comparison['Mô hình'])
plt.legend()
plt.ylim(0, 1.1)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# ============================================================================
# KẾT LUẬN
# ============================================================================
print("\n" + "="*60)
print("TÓM TẮT KẾT QUẢ CUỐI CÙNG")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# Tóm tắt kết quả
print("\n📊 THỐNG KÊ TỔNG QUAN:")
print(f"✓ Tổng số mẫu: {df.shape[0]}")
print(f"✓ Số thuộc tính ban đầu: {X.shape[1]}")
print(f"✓ Số thuộc tính sau one-hot encoding: {X_encoded.shape[1]}")
print(f"✓ Tỷ lệ train/test: 70%/30%")
print(f"✓ Số mẫu huấn luyện: {X_train.shape[0]}")
print(f"✓ Số mẫu kiểm thử: {X_test.shape[0]}")

print("\n🎯 KẾT QUẢ PHÂN LOẠI:")
print("-" * 80)
print(f"{'Mô hình':<20} {'Accuracy':<12} {'Precision':<12} {'Recall':<12} {'F1-Score':<12}")
print("-" * 80)
for idx, row in detailed_comparison.iterrows():
    print(f"{row['Mô hình']:<20} {row['Accuracy']:.4f}       {row['Precision']:.4f}       {row['Recall']:.4f}       {row['F1-Score']:.4f}")
print("-" * 80)

best_idx = comparison_df['Độ chính xác'].idxmax()
best_model = comparison_df.loc[best_idx, 'Mô hình']
best_accuracy = comparison_df.loc[best_idx, 'Độ chính xác']

print(f"\n🏆 MÔ HÌNH TỐT NHẤT: {best_model}")
print(f"   Độ chính xác: {best_accuracy:.4f} ({best_accuracy*100:.2f}%)")

print("\n💡 NHẬN XÉT:")
print("- Dataset Mushroom có ~8000 mẫu với 22-23 thuộc tính")
print("- Sau one-hot encoding, số thuộc tính tăng lên đáng kể")
print("- Cây quyết định (ID3 và CART) thường cho kết quả rất cao (>95%)")
print("- Naïve Bayes cũng hoạt động tốt nhờ thuộc tính rõ ràng")
print("- Với dữ liệu đầy đủ, các mô hình đều có độ chính xác cao")
print("- Thuộc tính 'odor' (mùi) thường là thuộc tính quan trọng nhất")

print("\n" + "="*60)
print("HOÀN THÀNH PHÂN TÍCH")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)
```
