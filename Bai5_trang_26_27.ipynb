```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.preprocessing import LabelEncoder
import warnings
warnings.filterwarnings('ignore')

# In thông tin sinh viên
print("="*60)
print("6351071057 - Lê Hoàng Phúc")
print("PHÂN TÍCH DỮ LIỆU HEART DISEASE")
print("="*60)

# ============================================================================
# YÊU CẦU A: NHẬP VÀ TIỀN XỬ LÝ DỮ LIỆU
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU A: NHẬP VÀ TIỀN XỬ LÝ DỮ LIỆU")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# Đọc dữ liệu
df = pd.read_csv('heart.csv')

print("\n1. Dữ liệu ban đầu:")
print("6351071057 - Lê Hoàng Phúc")
print(f"Kích thước: {df.shape}")
print(f"\nTên cột ban đầu:\n{df.columns.tolist()}")
print(f"\n5 dòng đầu tiên:\n{df.head()}")

# Kiểm tra dữ liệu trống
print("\n2. Kiểm tra dữ liệu trống:")
print("6351071057 - Lê Hoàng Phúc")
print(f"Tổng số giá trị null trong mỗi cột:")
print(df.isnull().sum())
print(f"\nTổng số dòng có giá trị null: {df.isnull().any(axis=1).sum()}")

# Xóa các dòng có dữ liệu trống (nếu có)
df_cleaned = df.dropna()
print(f"\nKích thước sau khi xóa dòng trống: {df_cleaned.shape}")
if df.shape[0] != df_cleaned.shape[0]:
    print(f"Đã xóa {df.shape[0] - df_cleaned.shape[0]} dòng")
else:
    print("Không có dòng nào bị xóa (không có dữ liệu trống)")

# Đổi tên các cột
print("\n3. Đổi tên các cột:")
print("6351071057 - Lê Hoàng Phúc")

new_column_names = {
    'age': 'age',
    'sex': 'sex',
    'cp': 'chest_pain_type',
    'trestbps': 'resting_blood_pressure',
    'chol': 'cholesterol',
    'fbs': 'fasting_blood_sugar',
    'restecg': 'rest_ecg',
    'thalach': 'max_heart_rate_achieved',
    'exang': 'exercise_induced_angina',
    'oldpeak': 'st_depression',
    'slope': 'st_slope',
    'ca': 'num_major_vessels',
    'thal': 'thalassemia',
    'target': 'target'
}

df = df_cleaned.rename(columns=new_column_names)

print(f"\nTên cột sau khi đổi:\n{df.columns.tolist()}")
print(f"\n5 dòng đầu tiên sau khi đổi tên:")
print(df.head())

# Thông tin dữ liệu
print("\n4. Thông tin dữ liệu sau tiền xử lý:")
print("6351071057 - Lê Hoàng Phúc")
print(f"\nKích thước: {df.shape}")
print(f"\nKiểu dữ liệu:\n{df.dtypes}")
print(f"\nThống kê mô tả:\n{df.describe()}")

# Phân bố nhãn target
print("\n5. Phân bố nhãn 'target':")
print("6351071057 - Lê Hoàng Phúc")
print(df['target'].value_counts())
print(f"\nTỷ lệ:")
print(df['target'].value_counts(normalize=True))

# Visualize target distribution
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
df['target'].value_counts().plot(kind='bar', color=['lightcoral', 'lightgreen'])
plt.title('Phân bố nhãn Target\n6351071057 - Lê Hoàng Phúc')
plt.xlabel('Target (0: Không bệnh, 1: Có bệnh)')
plt.ylabel('Số lượng')
plt.xticks(rotation=0)

plt.subplot(1, 2, 2)
df['target'].value_counts().plot(kind='pie', autopct='%1.1f%%',
                                   colors=['lightcoral', 'lightgreen'],
                                   labels=['Không bệnh tim (0)', 'Có bệnh tim (1)'])
plt.title('Tỷ lệ phân bố Target\n6351071057 - Lê Hoàng Phúc')
plt.ylabel('')

plt.tight_layout()
plt.show()
```

# Câu 3

```python
# ============================================================================
# YÊU CẦU 3: KHẢO SÁT ĐỘ TƯƠNG ĐỒNG (PEARSON)
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 3: KHẢO SÁT ĐỘ TƯƠNG ĐỒNG (PEARSON)")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# Tính ma trận tương quan Pearson
correlation_matrix = df.corr(method='pearson')
print("\nMa trận tương quan Pearson:")
print("6351071057 - Lê Hoàng Phúc")
print(correlation_matrix)

# Visualize correlation matrix
plt.figure(figsize=(14, 12))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Ma trận tương quan Pearson\n6351071057 - Lê Hoàng Phúc', fontsize=14)
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# Tìm các cặp cột có tương quan cao (> 0.7 hoặc < -0.7)
print("\nCác cặp cột có tương quan cao (|r| > 0.7):")
print("6351071057 - Lê Hoàng Phúc")
high_corr_pairs = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        if abs(correlation_matrix.iloc[i, j]) > 0.7 and correlation_matrix.columns[i] != 'target' and correlation_matrix.columns[j] != 'target':
            high_corr_pairs.append({
                'Column 1': correlation_matrix.columns[i],
                'Column 2': correlation_matrix.columns[j],
                'Correlation': correlation_matrix.iloc[i, j]
            })
            print(f"{correlation_matrix.columns[i]} - {correlation_matrix.columns[j]}: {correlation_matrix.iloc[i, j]:.3f}")

# Loại bỏ các cột có tương quan cao (không bao gồm target)
columns_to_drop = []
if len(high_corr_pairs) > 0:
    # Nếu có cặp tương quan cao, loại bỏ cột thứ 2 trong cặp
    for pair in high_corr_pairs:
        col_to_drop = pair['Column 2']
        if col_to_drop not in columns_to_drop and col_to_drop != 'target':
            columns_to_drop.append(col_to_drop)

    print(f"\nCác cột sẽ bị loại bỏ: {columns_to_drop}")
    df_reduced = df.drop(columns=columns_to_drop)
else:
    df_reduced = df.copy()
    print("\nKhông có cặp cột nào có tương quan > 0.7")
    print("Không loại bỏ cột nào.")

print(f"\nKích thước dữ liệu sau khi loại bỏ: {df_reduced.shape}")
print(f"Các cột còn lại: {df_reduced.columns.tolist()}")
```

# Câu 4

```python
# ============================================================================
# YÊU CẦU 4: TÁCH DỮ LIỆU
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 4: TÁCH DỮ LIỆU THÀNH THUỘC TÍNH VÀ NHÃN")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

X = df_reduced.drop('target', axis=1)  # Thuộc tính
y = df_reduced['target']  # Nhãn quyết định

print(f"\nKích thước thuộc tính X: {X.shape}")
print(f"Kích thước nhãn y: {y.shape}")
print(f"\nCác thuộc tính:\n{list(X.columns)}")
print(f"\nPhân bố nhãn target:")
print("6351071057 - Lê Hoàng Phúc")
print(y.value_counts().sort_index())
```

# Câu 5

```python
# ============================================================================
# YÊU CẦU 5: CHUYỂN ĐỔI ONE-HOT ENCODING
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 5: CHUYỂN ĐỔI ONE-HOT ENCODING")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# Kiểm tra các cột không phải số
non_numeric_cols = X.select_dtypes(include=['object']).columns.tolist()
print(f"\nCác cột không phải số: {non_numeric_cols if non_numeric_cols else 'Không có'}")

# Kiểm tra các cột category
categorical_cols = []
for col in X.columns:
    if X[col].nunique() <= 10:  # Giả sử cột có <= 10 giá trị unique là categorical
        categorical_cols.append(col)

print(f"\nCác cột có thể là categorical (≤10 giá trị unique):")
print("6351071057 - Lê Hoàng Phúc")
for col in categorical_cols:
    print(f"  {col}: {X[col].nunique()} giá trị unique - {sorted(X[col].unique())}")

if len(non_numeric_cols) > 0:
    X_encoded = pd.get_dummies(X, columns=non_numeric_cols, drop_first=True)
    print(f"\nKích thước sau one-hot encoding: {X_encoded.shape}")
    print(f"Các cột sau encoding: {list(X_encoded.columns)}")
else:
    X_encoded = X.copy()
    print("\nTất cả các cột đều là số.")
    print("Lưu ý: Một số cột có giá trị rời rạc nhưng đã được mã hóa sẵn (0, 1, 2, ...)")
    print("Không cần chuyển đổi thêm vì các thuật toán có thể xử lý được.")

print(f"\nKích thước cuối cùng của X: {X_encoded.shape}")
```

# Câu 6

```python
# ============================================================================
# YÊU CẦU 6: CHIA DỮ LIỆU TRAIN/TEST (70/30)
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 6: CHIA DỮ LIỆU TRAIN/TEST (70/30)")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.3, random_state=42, stratify=y
)

print(f"\nKích thước tập huấn luyện: {X_train.shape}")
print(f"Kích thước tập kiểm thử: {X_test.shape}")
print(f"\nPhân bố nhãn trong tập huấn luyện:")
print("6351071057 - Lê Hoàng Phúc")
print(y_train.value_counts().sort_index())
print(f"\nTỷ lệ: \n{y_train.value_counts(normalize=True).sort_index()}")
print(f"\nPhân bố nhãn trong tập kiểm thử:")
print(y_test.value_counts().sort_index())
print(f"\nTỷ lệ: \n{y_test.value_counts(normalize=True).sort_index()}")
```

# câu 7

```python
# ============================================================================
# YÊU CẦU 7: CÂY QUYẾT ĐỊNH ID3 (ENTROPY)
# ============================================================================
print("\n" + "="*60)
print("YÊU CẦU 7: CÂY QUYẾT ĐỊNH ID3 (ENTROPY)")
print("6351071057 - Lê Hoàng Phúc")
print("="*60)

# ID3 sử dụng entropy làm tiêu chí
id3_model = DecisionTreeClassifier(criterion='entropy', random_state=42, max_depth=5)
id3_model.fit(X_train, y_train)

# Dự đoán
y_pred_id3 = id3_model.predict(X_test)

# Ma trận nhầm lẫn
cm_id3 = confusion_matrix(y_test, y_pred_id3)
print("\nMa trận nhầm lẫn ID3:")
print("6351071057 - Lê Hoàng Phúc")
print(cm_id3)

# Độ chính xác
accuracy_id3 = accuracy_score(y_test, y_pred_id3)
print(f"\nĐộ chính xác ID3: {accuracy_id3:.4f} ({accuracy_id3*100:.2f}%)")

# Classification report
print("\nBáo cáo phân loại ID3:")
print("6351071057 - Lê Hoàng Phúc")
print(classification_report(y_test, y_pred_id3, target_names=['Không bệnh (0)', 'Có bệnh (1)']))

# Visualize confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm_id3, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Không bệnh (0)', 'Có bệnh (1)'],
            yticklabels=['Không bệnh (0)', 'Có bệnh (1)'])
plt.title('Ma trận nhầm lẫn - Cây ID3\n6351071057 - Lê Hoàng Phúc')
plt.ylabel('Nhãn thực tế')
plt.xlabel('Nhãn dự đoán')
plt.tight_layout()
plt.show()

# Visualize tree
plt.figure(figsize=(20, 12))
plot_tree(id3_model, feature_names=X_encoded.columns,
          class_names=['Không bệnh', 'Có bệnh'],
          filled=True, rounded=True, fontsize=9)
plt.title('Cây quyết định ID3\n6351071057 - Lê Hoàng Phúc', fontsize=14)
plt.tight_layout()
plt.show()

# Feature importance
feature_importance_id3 = pd.DataFrame({
    'Feature': X_encoded.columns,
    'Importance': id3_model.feature_importances_
}).sort_values('Importance', ascending=False)

print("\nĐộ quan trọng của các thuộc tính (ID3) - Top 10:")
print("6351071057 - Lê Hoàng Phúc")
print(feature_importance_id3.head(10))

# Visualize feature importance
plt.figure(figsize=(10, 6))
top_features = feature_importance_id3.head(10)
plt.barh(top_features['Feature'], top_features['Importance'], color='skyblue')
plt.xlabel('Độ quan trọng')
plt.title('Top 10 thuộc tính quan trọng nhất (ID3)\n6351071057 - Lê Hoàng Phúc')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
